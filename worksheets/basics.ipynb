In order to compute in the symmetric functions we need to define at minimum a basis.  In the command below we define the Schur basis:

```python
from sage.all import *
```

```python
s = SymmetricFunctions(QQ).s()
```

With it we can compute plethysms, however for degree > 30 it will take a long time.

```python
s[5](s[3])
```

```python
s[5](s[3])
```

The series we are computing have the length of the partitions restricted, but the plethysms can have terms with longer lengths.  If we want to just focus on the terms that have a length less than or equal to three we can sum over all terms $(\lambda, c)$ and add up only those that satisfy $\ell(\lambda)\leq 3$.

```python
sum(c*s(la) for (la,c) in s[5](s[3]) if len(la)<=3)
```

We know that $s_\lambda[s_\mu]$ will have all terms of length less than or equal to $|\lambda|\ell(\mu)$.

```python
s[3](s[5])
```
